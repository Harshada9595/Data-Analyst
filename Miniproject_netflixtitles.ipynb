DATA VISUALIZATION

```python
df["type"].value_counts().plot(kind="bar")
plt.title("Movies Vs TV show")
plt.xlabel("Type")
plt.ylabel("Count")
plt.show()
```

Content by year

```python
year_count = df["type"].value_counts().sort_index()
plt.figure(figsize=(10, 12))
plt.plot(year_count.index, year_count.values)
plt.title("Netflix content by realasing year")
plt.xlabel("Year")
plt.ylabel("Number of titles")
plt.show()
```
